# CHECKPOINT 1 — Problem Framing

Business Problem:

The goal of this project is to build a machine learning model that can predict real-estate property prices based on available property characteristics. This could help estimate reasonable property prices and support better decision-making.

Target Variable: 

The target variable is `price`, which represents the property's listing price in AZN.

Success Metrics:

We will use:
- MAE (Mean Absolute Error) as the main metric
- RMSE (Root Mean Squared Error)
- R² (R-squared)

MAE will be the main metric because it shows the average prediction error in AZN and is easy to interpret.

Machine Learning Task:

This is a supervised learning regression problem because we are predicting a continuous numerical value (`price`).

# CHECKPOINT 2 — Full EDA and Cleaning

In this step, we explore the dataset, identify data quality issues, remove unnecessary or duplicated information, handle missing values and outliers, and prepare the data for modeling.

```python
#2.1 Initial Dataset Inspection

# First, we load the dataset and examine its basic structure, available variables, data types, missing values, duplicate rows, and descriptive statistics. This helps us understand the dataset before performing further cleaning and analysis.

import pandas as pd

# firstly we load the dataset
df = pd.read_csv("house_sale.csv")

# the number of rows and columns
print("Shape:", df.shape)

# which variables are available?
print("\nColumns:")
print(df.columns.tolist())

# data type of all columns
print("\nData types:")
print(df.dtypes)

# missing values
print("\nMissing values:")
print(df.isnull().sum())

# duplicated rows
print("\nDuplicates:", df.duplicated().sum())

print("\nFirst 5 rows:")
display(df.head())

# to udnerstand data more clearly: create descriptive statistics 
print("\nNumerical summary:")
display(df.describe())
```

```python
# 2.2 Checking `price` and `total_price`

# Based on the initial inspection, we noticed that `price` and `total_price` contain the same number of values. We therefore check whether they are actually identical. If they are, `total_price` should not be used as a feature because it duplicates our target variable `price` and would cause data leakage.

print("price == total_price:",
      (df["price"] == df["total_price"]).mean())

# the differences between the two columns
print("\nDifference between price and total_price:")
print((df["price"] - df["total_price"]).describe())
```

```python
# so, conclusion:  0.9998709997519226 is showing that the rows are identical
```

```python
# 2.3 Checking `_x` and `_y` Columns

# The dataset contains several pairs of `_x` and `_y` columns, which likely came from merging two datasets. We compare each pair to determine whether they contain the same information. If two columns are identical, we can remove one of them because keeping both would be redundant.
# I'll check whether they contain the same information, in order to avoid keeping duplicate features

pairs = [
    ("id_x", "id_y"),
    ("estate_rel_url_x", "estate_rel_url_y"),
    ("datetime_scrape_x", "datetime_scrape_y"),
    ("currency_x", "currency_y"),
    ("estate_details_id_x", "estate_details_id_y")
]

for col1, col2 in pairs:
    same = (df[col1] == df[col2]).mean()
    print(f"{col1} vs {col2}: {same:.2%} identical")
```

```python
# _x and _y check shows that 3 pairs are duplicates, but the other 2 pairs are different
```

```python
# 2.4 Analyzing Extreme Price Values

# The initial analysis showed unusually low and high property prices
# to see if they they are real properties or data errors:
# To check:

print("Prices below 1,000 AZN:")
display(df[df["price"] < 1000][
    ["price", "location", "attributes", "Sahə", "Otaq sayı"]
].head(20))

print("\nPrices above 10 million AZN:")
display(df[df["price"] > 10_000_000][
    ["price", "location", "attributes", "Sahə", "Otaq sayı"]
].head(20))

# how many extreme price values are there?

print("Below 1,000 AZN:", (df["price"] < 1000).sum())
print("Above 10 million AZN:", (df["price"] > 10_000_000).sum())

# Check whether extreme prices have reasonable price per m²?

df["price_per_m2"] = df["price"] / df["Sahə"].str.extract(r"([\d.]+)")[0].astype(float)

display(
    df[["price", "Sahə", "price_per_m2"]]
    .sort_values("price_per_m2", ascending=False)
    .head(20)
)
```

```python
# 2.5 Checking for Repeated Property Listings

# The dataset may contain the same property multiple times because listings can be in different scrapings. We check the property ID and URL to find repeated listing

# Check for repeated property listings

print("Duplicate estate IDs:", df["estate_id"].duplicated().sum())

print("Duplicate URLs:", df["estate_rel_url"].duplicated().sum())
```

```python
# 2.7 Removing Leakage and Non-Predictive Columns

# Now we go back to the previous analysis  where we checked the duplicated and suspicious columns first
# We found that total_price is essentially identical to our target price 
# so keeping it would cause data leakage. then we also found that the _x/_y and ID/URL columns are identifiers or duplicated information 
# rather than useful property features
# Therefore, we decided to remove these columns before modeling.
```

```python
drop_cols = [
    "total_price",          # same information as price -> leakage
    "estate_rel_url",       # URL/identifier
    "estate_rel_url_x",     # URL/identifier
    "estate_rel_url_y",     # URL/identifier
    "id_x",                 # identifier
    "id_y",                 # identifier
    "estate_id",            # identifier
    "estate_details_id_x",  # identifier
    "estate_details_id_y",  # identifier
    "rel_url",              # scraping URL
    "img_url"               # image URL
]

df_model = df.drop(columns=drop_cols)

print("Shape after dropping:", df_model.shape)
```

```python
# lets check missing values after removing leakage and identifier columns
missing_percent = (df_model.isnull().mean() * 100).sort_values(ascending=False)

print("Missing values (%) in remaining columns:")
print(missing_percent)
```

```python
# Removing Features with Excessive Missing Values
# we see that, some columns have too much missing data, it is better to remove them

drop_missing_cols = [
    "Binanın növü",
    "featured",
    "vip",
    "Torpaq sahəsi",
    "hour_y"
]

df_clean = df_model.drop(columns=drop_missing_cols)

print("Shape after removing highly incomplete features:", df_clean.shape)
```

```python
# Checking Duplicate Features
# then, İpoteka and mortgage both had exactly 67.31% missing data, now we need to check them to see if they are dublicates

same = (df_clean["İpoteka"] == df_clean["mortgage"]).mean()

print(f"İpoteka vs mortgage: {same:.2%} identical")
```

```python
# as we can see : Mərtəbə and Sahə are stored as text, we need to convert them to numbers

# Converting Property Features to Numerical Values
df_clean["area_m2"] = (
    df_clean["Sahə"]
    .str.extract(r"([\d.]+)")[0]
    .astype(float)
)

# Extract current floor from Mərtəbə
df_clean["floor"] = (
    df_clean["Mərtəbə"]
    .str.extract(r"^(\d+)")[0]
    .astype(float)
)

# Extract total number of floors
df_clean["total_floors"] = (
    df_clean["Mərtəbə"]
    .str.extract(r"/\s*(\d+)")[0]
    .astype(float)
)

print(df_clean[[
    "Sahə",
    "area_m2",
    "Mərtəbə",
    "floor",
    "total_floors"
]].head(10))
```
